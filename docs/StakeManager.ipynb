```python
# Define time constants
T_MINUTE = var('T_MINUTE') #60
T_HOUR = var('T_HOUR') #60 * T_MINUTE
T_DAY = var('T_DAY') #24 * T_HOUR
T_WEEK = var('T_WEEK') #7 * T_DAY
T_YEAR = var('T_YEAR') #365 * T_DAY + 5 * T_HOUR + 48 * T_MINUTE + 45;

# Define system constants
SCALE_FACTOR = 10^18
M_MAX = var('M_MAX') #4
APY = var('APY') # 100  # Annual Percentage Yield (percent)
MPY = M_MAX * APY
MPY_abs = 100 + (2 * M_MAX * APY)
T_RATE = var('T_RATE') #1 * T_WEEK  # seconds
A_MIN = ceil((T_YEAR * 100) / (T_RATE * APY))  # Minimal tokens balance
A_MAX = (2^256 - 1) # Maximum token balance 
T_MIN = 90 * T_DAY  # Minimum lockup period in seconds
T_MAX = M_MAX * T_YEAR  # Maximum lockup period in seconds

# Define variables
mp, mp_M, mp_Sigma, mp_target, a_bal = var('mp mp_M mp_Sigma mp_target a_bal')
t_lock, t_lock_end, t_now = var('t_lock t_lock_end t_now')
Delta_t, Delta_a = var('Delta_t Delta_a')
S_start, E_num, E_current = var('S_start E_num E_current')


mp_Sigma_acc = min((a_bal * T_RATE * APY) / (100 * T_YEAR), mp_M - mp_Sigma)
t_lock_rem = max(t_lock_end, t_now) - t_now
```

#### Initial Multiplier Points
\begin{equation}
    \mathcal{f}{mp_\mathcal{I}}(\Delta a) = \Delta a
\end{equation}

```python
mp_I = a_bal
show(mp_I)
```

#### Accrued Multiplier Points
\begin{equation}
    \mathcal{f}mp_\mathcal{A}(a_{bal}, \Delta t) = \dfrac{a_{bal} \times \Delta t \times \mathtt{APY}}{100 \times T_{YEAR}}
\end{equation}

```python
mp_A = (a_bal * Delta_t * APY) / (100 * T_YEAR)
show(mp_A)
```

#### Bonus Multiplier Points
\begin{equation}
    \mathcal{f}mp_\mathcal{B}(\Delta a, t_{lock})  = \dfrac{\Delta a \times t_{lock} \times \mathtt{APY}}{100 \times T_{YEAR}}
\end{equation}

```python
mp_B = (a_bal * t_lock * APY) / (100 * T_YEAR)
show(mp_B)
```

#### Reduce Multiplier Points
\begin{equation}
    \mathcal{f}{mp_\mathcal{R}}(mp, a_{bal}, \Delta a) = \dfrac{mp \times \Delta a}{ a_{bal}}
\end{equation}

```python
mp_R = (mp * Delta_a) / a_bal
show(mp_R)
```

#### Maximum Total Multiplier Points
\begin{equation}
    \hat{\mathcal{f}}mp_{\mathcal{M}}(a_{bal}, t_{\text{lock}}) = a_{bal} + \frac{a_{bal} \times \mathtt{APY} \times \left( M_{\text{MAX}} \times T_{\text{YEAR}} + t_{\text{lock}} \right)}{100 \times T_{\text{YEAR}}}
\end{equation}

```python
f_mp_M = a_bal + (a_bal * APY * (M_MAX * T_YEAR + t_lock)) / (100 * T_YEAR)
show(f_mp_M)
```

```python
f_mp_Sigma = mp_I + mp_B + mp_A
show(f_mp_Sigma)
```

#### Maximum Bonus Multiplier Points


```python
mp_B_max = (a_bal * T_MAX * APY) / (100 * T_YEAR)
show(mp_B_max)
```

#### Maximum Accrued Multiplier Points
\begin{equation}
    \hat{\mathcal{f}}mp_\Sigma^{max}(a_{bal}) = \frac{a_{bal} \times \mathsf{MPY}}{100}
\end{equation}

```python
mp_A_max = (a_bal * MPY) / 100
show(mp_A_max)
```

#### Maximum Absolute Multiplier Points
\begin{equation}
    \hat{\mathcal{f}}mp_\mathcal{M}^\mathit{abs}(a_{bal}) = \frac{a_{bal} \times \mathsf{MPY}^\mathit{abs}}{100}
\end{equation}

```python
mp_M_abs = (a_bal * MPY_abs) / 100
show(mp_M_abs)
```

#### Time to Accrue Multiplier Points
\begin{equation}
    t_{rem}(a_{bal},mp) = \frac{mp \times 100 \times T_{YEAR}}{a_{bal} \times \mathtt{APY}}
\end{equation}

```python
t_rem = (mp * 100 * T_YEAR) / (a_bal * APY)
show(t_rem)
```

#### Time to Accrue Maximum Multiplier Points
\begin{equation}
	t_{rem}^{max}(a_{bal},mp_\Sigma,mp_\mathcal{M}) = \frac{(mp_\mathcal{M} - mp_\Sigma) \times 100 \times T_{YEAR}}{a_{bal} \times \mathtt{APY}}
\end{equation}

```python
t_rem_max = ((mp_M - mp_Sigma) * 100 * T_YEAR) / (a_bal * APY)
show(t_rem_max)
```

#### Locked Time Retrival
\begin{equation}
		\hat{\mathcal{f}}\tilde{t}_{lock}(mp_{\mathcal{M}}, a_{bal}) \approx \left\lceil \frac{(mp_{\mathcal{M}} - a_{bal}) \times 100 \times T_{YEAR}}{a_{bal} \times \mathtt{APY}}\right\rceil - T_{\text{YEAR}} \times M_{\text{MAX}}
\end{equation}

```python
t_lock_est = ceil(((mp_M - a_bal) * 100 * T_YEAR) / (a_bal * APY)) - (M_MAX * T_YEAR)
show(t_lock_est)
```

#### Remaining Time Lock Allowed to Increase
\begin{equation}
    t_{rem}^{lock}(a_{bal},mp_\mathcal{M}) \approx \frac{(\frac{a_{bal} \times \mathsf{MPY}^\mathit{abs}}{100} - mp_\mathcal{M})\times T_{YEAR}}{a_{bal}}
\end{equation}

```python
t_rem_lock = ((((a_bal * MPY_abs)/100)-mp_M) * T_YEAR) / a_bal
show(t_rem_lock)
```

#### MP per Accrual Rate
\begin{equation}
	mp_{\text{rate}}(a_{bal}) = \frac{a_{bal} \times T_{RATE} \times \mathtt{APY}}{100 \times T_{YEAR}}
\end{equation}

```python
mp_rate = (a_bal * T_RATE * APY) / (100 * T_YEAR)
show(mp_rate)
```

#### Epoch Start Time
\begin{equation}
	E_{start}(E_{num}) = S_{start} + (T_{RATE} \times E_{num})
\end{equation}

```python
E_start = S_start + (T_RATE * E_num)
show(E_start)
```

#### MP Accrued for Remaining Time in Current Epoch
\begin{equation}
    \Delta t_{\text{epoch}} = E_{start}(E_{current} +1) - t_{now}
\end{equation}
\begin{equation}
	mp_{\text{fractional}} = mp_{\text{rate}} - \frac{a_{bal} \times \Delta t_{\text{epoch}} \times \mathtt{APY}}{100 \times T_{YEAR}}
\end{equation}	

```python
Delta_t_epoch = S_start + (T_RATE * E_current) - t_now
#show(Delta_t_epoch)

mp_epoch_remaining = ((a_bal * Delta_t_epoch * APY)/(100*T_YEAR))
 #show(mp_epoch_remaining)
mp_rate = var('mp_rate')
mp_fractional = mp_rate - mp_epoch_remaining
show(mp_fractional)
```

\begin{equation}
	mp_{\text{target}}(a_{bal}) = \frac{a_{bal} \times \mathsf{MPY}}{100} + mp_{\text{fractional}}
\end{equation}

```python
mp_fractional = var('mp_fractional')
mp_target = mp_A_max + mp_fractional
show(mp_target)
```

\begin{equation}
	\Delta E_{\text{target,1}} = \left\lfloor \frac{mp_{\text{target}}}{mp_{\text{rate}}} \right\rfloor
\end{equation}

```python
delta_e_target=floor(mp_target / mp_rate)
#show(delta_e_target)

epoch_target = E_current + delta_e_target
show(epoch_target)
```

\begin{equation}
	\Delta t_{rem}^{mp_M} = \frac{(mp_\mathcal{M} - mp_\Sigma ) \times T_{RATE}}{\frac{a_{bal} \times T_{RATE} \times \mathtt{APY}}{100 \times T_{YEAR}}}\\
\end{equation}

```python
delta_t_mp_M_rem = ((mp_M - mp_Sigma) * T_RATE) / ((a_bal*T_RATE*APY)/(100*T_YEAR))
show(delta_t_mp_M_rem)
```
